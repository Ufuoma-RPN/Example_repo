```python
my_dict = {"Continents": ["Africa", "North Ameria", "South America", "Asia", "Europe"], "Country_names": ["Egypt", "Canada", "Argentina", "Taiwan", "Germany"], "Population": [109300000, 38250000, 45810000, 23570000, 83200000]}
my_dataframe = pd.DataFrame(my_dict)
my_dataframe
```
