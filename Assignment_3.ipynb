**1. Python function to calcuate the factorial of a number.**

```python
def fact(n):
  return 1 if (n == 1 or n == 0) else n * fact(n-1);
n = int(input())
print("Factorial of", n, "is")
fact(n)
```

**2. A function that takes a list of numbers as input and returns the sum of the numbers in the list as output.**

```python
list_1 = input().split() #This is to split the input string into a list of strings using the comma
list_1 = list(map(int, list_1)) #This is to convert the strings to integers.
sum = 0
for value in list_1:
  sum = sum + value
print(sum)
```

**3. Define a function that takes a string as input and returns the string reversed.**

```python
platform = str(input()) [::-1]
print(platform)
```
