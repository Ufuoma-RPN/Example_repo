```python
# Load the workbook with all sheets
workbook = pd.ExcelFile(filename)  # Replace with your file name
```

```python
# Iterate through each sheet and create a data dictionary
data_dicts = {}

for sheet_name in workbook.sheet_names:
    df = workbook.parse(sheet_name)  # Load the sheet into a DataFrame
    data_dict = pd.DataFrame({
        "Column Name": df.columns,
        "Data Type": [df[col].dtype for col in df.columns],
        "Non-Null Count": [df[col].count() for col in df.columns],
        "Unique Values": [df[col].nunique() for col in df.columns],
        "Sample Value": [df[col].iloc[0] if not df[col].empty else None for col in df.columns]
    })
    data_dicts[sheet_name] = data_dict
```

```python
for sheet_name, data_dict in data_dicts.items():
    file_name = f'{sheet_name}_data_dictionary.csv'
    data_dict.to_csv(file_name, index=False)
    files.download(file_name)
```

```python
with pd.ExcelWriter('data_dictionaries.xlsx') as writer:
    for sheet_name, data_dict in data_dicts.items():
        data_dict.to_excel(writer, sheet_name=sheet_name, index=False)

# Download the combined file
files.download('data_dictionaries.xlsx')
```
